```python
# Prompt user for Input. Ask the user to enter a number for which they want to
# generate a multiplication table.
number = int(input('Enter a number: '))
# Generate the Multiplication Table: Use a for loop to iterate through the numbers
# 1 to 10. In each iteration, calculate the product of the user's number and the
# current number from the loop.
for i in range(1,11):
    result = number*i
    print(number,' x ',i,'=',result)
    
```
